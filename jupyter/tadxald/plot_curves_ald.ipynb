```python
def plot_uq_(iterL, bestA, color, label, optval=None, optgap=None,
            fignum=None):
   
    resA = np.array(bestA)

    low, mid, high = np.percentile(resA.T, [2.5, 50, 97.5], axis=1)


    if optgap is None:
        plt.plot(iterL, mid, alpha=0.9, color=color, label=label,linewidth = 5)
        plt.plot(iterL, low, alpha=0.9, ls=':', color=color,linewidth = 5)
        plt.plot(iterL, high, alpha=0.9, ls=':', color=color,linewidth = 5)
        plt.fill_between(
            iterL, low, high, alpha=0.15, facecolor=color)
        plt.xlabel('ALD cycles')
        
        plt.xlim(0, 840)
        plt.legend()
    else:
        low = low - optval
        mid = mid - optval
        high = high - optval
        plt.semilogy(iterL, mid, alpha=0.9, color=color, label=label)
        plt.semilogy(iterL, low, alpha=0.9, ls=':', color=color)
        plt.semilogy(iterL, high, alpha=0.9, ls=':', color=color)
        plt.fill_between(
            iterL, low, high, alpha=0.15, facecolor=color)
```
